## 가위바위보 게임

사용자 vs 컴퓨터
if/elif로 승패 판정
while로 여러 판 진행
3선승제

1. 가위바위보 선택지 만들기
2. 사용자 입력받기
3. 입력값 검사
4. 승패 판정
5. 한 판이 끝나면 다시 진행
6. 승리 횟수 추가
7. 3선승제
8. 최종 승자 판정

```python
import random

# 1. 가위바위보 선택지 만들기
choices = ["가위", "바위", "보"]

count = 0 # 최종 게임 진행 횟수

# 이길 경우 딕셔너리(key값은 사용자가 낸 것)
win = {
    "가위" : "보",
    "바위" : "가위",
    "보" : "바위"
    }

user_win = 0 # 사용자가 이긴 횟수 저장 변수
computer_win = 0 # 컴퓨터가 이긴 횟수 저장 변수

print("가위바위보 게임 3선승제!")

# 2. 3선승제 시작
while user_win < 3 and computer_win <3:

    # 3. 컴퓨터의 랜덤 선택
    computer_choice = random.choice(choices)

    # 4. 사용자 입력값 받기
    user_choice = input("가위, 바위, 보 중 하나 입력하세요! : ")

    # 5. 사용자가 가위, 바위, 보 말고 다른 단어를 입력했을 때 출력
    if user_choice not in choices: 
        print("(가위, 바위, 보) 셋 중 하나 고르세요")
        continue

    # 6. 사용자의 선택과 컴퓨터의 선택 출력
    print(f"사용자 : {user_choice}")
    print(f"컴퓨터 : {computer_choice}")


    # 한번 진행 될 때마다 +1
    count += 1


    # 7. 가위바위보 진행
    if user_choice == computer_choice:
        print("비겼습니다!")

    elif win[user_choice] == computer_choice:
        print("이겼습니다!")
        user_win += 1 # 사용자가 이겼으므로 user_win 증가

    else:
        print("졌습니다ㅜㅜ")
        computer_win += 1 # 사용자가 졌으므로 computer_win 증가

    # 8. 현재 스코어 
    print(f"현재 스코어 -> 사용자 {user_win} : {computer_win} 컴퓨터")
    print("========================================")

# 9. 3선승제 확인
if user_win == 3: #사용자가 3번 이겼을 경우 게임 종료
    print("🎉 최종 승리를 축하합니다!")
    print(f"총 횟수 : {count}")

else: # 컴퓨터가 3번 이겼을 경우 게임 종료
    print("컴퓨터가 이겼습니다ㅜㅜ")
```
