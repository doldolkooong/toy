```python
import random

#1. 랜덤 숫자 만들기
answer = random.randint(1, 100)
count = 0

print("<숫자 맞추기 게임> \n 1~100 사이의 숫자를 맞춰보세요!")

#2. 사용자에게 숫자 입력 받기

while True:
    user = int(input("숫자를 입력하세요: "))

    #3. 1~100 사이의 숫자를 입력하도록 하는 문구 출력
    if user > 100:
        print("1~100 사이의 숫자를 입력해주세요.")
        continue
    
    #4. 정답을 맞추기 위해 시도했던 횟수 확인
    count += 1

    #5. 정답보다 크고 작은지 판단
    if user > answer: # 정답보다 클 경우
        print("========================")
        print(f"입력한 숫자 : {user}")
        print("더 작은 숫자입니다.")

    elif user < answer: # 정답보다 작을 경우
        print("========================")
        print(f"입력한 숫자 : {user}")
        print("더 큰 숫자입니다.")

    else: # 정답일 경우
        print("========================")
        print(f"숫자 입력 : {user}")
        print("🎉 정답입니다!")
        print(f"{count}번 만에 맞혔습니다!")
        break # 정답일 경우 반복 종료

```
